# I used Python and Pandas to clean and validate IT support ticket data. I handled missing categories, standardized inconsistent category values, removed duplicate tickets, corrected invalid timestamps, and performed data-quality validation checks.

```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

np.random.seed(42)
random.seed(42)

print("Libraries loaded successfully")
```

```python
teams = [
    "Service Desk",
    "Network",
    "Application Support",
    "Infrastructure",
    "Database"
]

categories = [
    "Password Reset",
    "VPN",
    "Application Error",
    "Network Connectivity",
    "Hardware",
    "Email",
    "Access Request",
    "Database Issue"
]

priorities = [
    "Low",
    "Medium",
    "High",
    "Critical"
]

statuses = [
    "Resolved",
    "Open",
    "In Progress",
    "Pending"
]

request_types = [
    "Incident",
    "Service Request"
]

agents = [
    "Agent 01",
    "Agent 02",
    "Agent 03",
    "Agent 04",
    "Agent 05",
    "Agent 06",
    "Agent 07",
    "Agent 08",
    "Agent 09",
    "Agent 10"
]

print("Lists created successfully")
```

```python
#Create 3,500 Ticket IDs
num_tickets = 3500

ticket_ids = [
    f"INC{i:05d}"
    for i in range(1, num_tickets + 1)
]

print("Number of tickets:", len(ticket_ids))
print("First 5:", ticket_ids[:5])
print("Last 5:", ticket_ids[-5:])
```

```python
#Generate created dates
start_date = datetime(2025, 1, 1)
end_date = datetime(2026, 8, 31)

date_range_days = (end_date - start_date).days

created_dates = [
    start_date + timedelta(
        days=random.randint(0, date_range_days),
        hours=random.randint(0, 23),
        minutes=random.randint(0, 59)
    )
    for _ in range(num_tickets)
]

print("First created date:", created_dates[0])
print("Last created date:", created_dates[-1])
```

```python
#Generate the ticket attributes
ticket_teams = np.random.choice(
    teams,
    size=num_tickets,
    p=[0.35, 0.20, 0.20, 0.15, 0.10]
)

ticket_categories = np.random.choice(
    categories,
    size=num_tickets,
    p=[0.16, 0.15, 0.14, 0.13, 0.10, 0.10, 0.12, 0.10]
)

ticket_priorities = np.random.choice(
    priorities,
    size=num_tickets,
    p=[0.20, 0.45, 0.25, 0.10]
)

ticket_statuses = np.random.choice(
    statuses,
    size=num_tickets,
    p=[0.70, 0.10, 0.15, 0.05]
)

ticket_request_types = np.random.choice(
    request_types,
    size=num_tickets,
    p=[0.75, 0.25]
)

ticket_agents = np.random.choice(
    agents,
    size=num_tickets
)

print("Ticket attributes generated successfully")
```

```python
#Create SLA targets
sla_targets = {
    "Low": 48,
    "Medium": 24,
    "High": 8,
    "Critical": 4
}

sla_target_hours = [
    sla_targets[priority]
    for priority in ticket_priorities
]

print(sla_target_hours[:10])
```

```python
#Generate resolution times
resolution_hours = []

for priority in ticket_priorities:

    if priority == "Critical":
        hours = np.random.gamma(shape=2, scale=2)

    elif priority == "High":
        hours = np.random.gamma(shape=2.5, scale=3)

    elif priority == "Medium":
        hours = np.random.gamma(shape=2.5, scale=5)

    else:
        hours = np.random.gamma(shape=2.5, scale=8)

    resolution_hours.append(round(hours, 2))

print(resolution_hours[:10])
```

```python
#Create the DataFrame
df = pd.DataFrame({
    "Ticket_ID": ticket_ids,
    "Created_Date": created_dates,
    "Team": ticket_teams,
    "Category": ticket_categories,
    "Priority": ticket_priorities,
    "Status": ticket_statuses,
    "Request_Type": ticket_request_types,
    "Assigned_Agent": ticket_agents,
    "SLA_Target_Hours": sla_target_hours,
    "Resolution_Hours": resolution_hours
})

df.head()
```

```python
resolved_dates = []

for status, created, resolution in zip(
    df["Status"],
    df["Created_Date"],
    df["Resolution_Hours"]
):
    
    if status == "Resolved":
        resolved_date = created + timedelta(hours=resolution)
    else:
        resolved_date = pd.NaT
    
    resolved_dates.append(resolved_date)

df["Resolved_Date"] = resolved_dates
```

```python
df[[
    "Ticket_ID",
    "Created_Date",
    "Resolved_Date",
    "Status",
    "Resolution_Hours"
]].head(10)
```

```python
df["SLA_Status"] = np.where(
    df["Resolution_Hours"] <= df["SLA_Target_Hours"],
    "Met",
    "Breached"
)
```

```python
df["SLA_Status"].value_counts()
```

```python
df.loc[
    df["Status"] != "Resolved",
    "Resolution_Hours"
] = np.nan
```

```python
#Recalculate SLA Status
#Because we changed Resolution_Hours, we should recalculate SLA_Status.

df["SLA_Status"] = np.where(
    df["Resolution_Hours"].isna(),
    "Not Resolved",
    np.where(
        df["Resolution_Hours"] <= df["SLA_Target_Hours"],
        "Met",
        "Breached"
    )
)
```

```python
df["SLA_Status"].value_counts()
```

```python
#Add Reopened
#A ticket can be resolved and later reopened.

reopened = []

for category, status in zip(
    df["Category"],
    df["Status"]
):
    
    if status == "Resolved":
        
        if category in [
            "VPN",
            "Application Error",
            "Network Connectivity",
            "Database Issue"
        ]:
            probability = 0.12
        else:
            probability = 0.05
        
        value = np.random.choice(
            ["Yes", "No"],
            p=[probability, 1 - probability]
        )
        
    else:
        value = "No"
    
    reopened.append(value)

df["Reopened"] = reopened
```

```python
df["Reopened"].value_counts()
```

```python
#Add Customer_Satisfaction

satisfaction_scores = []

for status in df["Status"]:
    
    if status == "Resolved":
        score = np.random.choice(
            [1, 2, 3, 4, 5],
            p=[0.05, 0.10, 0.20, 0.40, 0.25]
        )
    else:
        score = np.nan
    
    satisfaction_scores.append(score)

df["Customer_Satisfaction"] = satisfaction_scores
```

```python
#Add missing Category values
#We'll intentionally remove the category for 30 tickets.

missing_category_indices = np.random.choice(
    df_raw.index,
    size=30,
    replace=False
)

df_raw.loc[missing_category_indices, "Category"] = np.nan

print("Missing Category values:", df_raw["Category"].isna().sum())
```

```python
#Add inconsistent Category names

category_changes = {
    "VPN": ["vpn", "VPN Issue"],
    "Password Reset": ["password reset", "Password reset"],
    "Email": ["email", "Email Issue"],
    "Hardware": ["hardware", "Hardware Issue"]
}

for original, variations in category_changes.items():
    
    indices = df_raw[
        df_raw["Category"] == original
    ].sample(
        n=min(10, (df_raw["Category"] == original).sum()),
        random_state=42
    ).index
    
    for i, variation in zip(indices, variations * 10):
        df_raw.loc[i, "Category"] = variation
```

```python
df_raw["Category"].value_counts(dropna=False)
```

```python
#Add duplicate tickets
duplicate_rows = df_raw.sample(
    n=20,
    random_state=42
)

df_raw = pd.concat(
    [df_raw, duplicate_rows],
    ignore_index=True
)

print("New shape:", df_raw.shape)
print("Duplicate rows:", df_raw.duplicated().sum())
```

```python
#Add missing Resolved_Date values
#For some resolved tickets, we'll remove the resolution timestamp.

resolved_indices = df_raw[
    df_raw["Status"] == "Resolved"
].sample(
    n=40,
    random_state=42
).index

df_raw.loc[
    resolved_indices,
    "Resolved_Date"
] = pd.NaT

print(
    "Missing Resolved_Date:",
    df_raw["Resolved_Date"].isna().sum()
)
```

```python
#Add invalid dates
#We'll intentionally make 10 resolved dates earlier than their created dates.

invalid_indices = df_raw[
    df_raw["Status"] == "Resolved"
].sample(
    n=10,
    random_state=42
).index

df_raw.loc[
    invalid_indices,
    "Resolved_Date"
] = (
    df_raw.loc[invalid_indices, "Created_Date"]
    - pd.to_timedelta(2, unit="hours")
)

print("Invalid dates introduced:", len(invalid_indices))
```

```python
#Add missing Customer Satisfaction
#We'll make 25 resolved tickets have missing satisfaction scores.

satisfaction_indices = df_raw[
    df_raw["Status"] == "Resolved"
].sample(
    n=25,
    random_state=42
).index

df_raw.loc[
    satisfaction_indices,
    "Customer_Satisfaction"
] = np.nan

print(
    "Missing satisfaction:",
    df_raw["Customer_Satisfaction"].isna().sum()
)
```

```python
#Check our raw data

print("Shape:", df_raw.shape)

print("\nMissing values:")
print(df_raw.isna().sum())

print("\nDuplicate rows:")
print(df_raw.duplicated().sum())
```

```python
#Start Cleaning
#We'll clean this professionally, one issue at a time.

#Step 5.1 — Convert date columns

df_raw["Created_Date"] = pd.to_datetime(
    df_raw["Created_Date"],
    errors="coerce"
)

df_raw["Resolved_Date"] = pd.to_datetime(
    df_raw["Resolved_Date"],
    errors="coerce"
)
```

```python
df_raw[["Created_Date", "Resolved_Date"]].dtypes
```

```python
#Check the inconsistent categories
#This is exactly the type of issue a Data Analyst needs to clean.

df_raw["Category"].value_counts(dropna=False)
```

```python
#Standardize the categories

category_mapping = {
    "vpn": "VPN",
    "VPN Issue": "VPN",
    "password reset": "Password Reset",
    "Password reset": "Password Reset",
    "email": "Email",
    "Email Issue": "Email",
    "hardware": "Hardware",
    "Hardware Issue": "Hardware"
}

df_raw["Category"] = df_raw["Category"].replace(category_mapping)
```

```python
#The inconsistent names should now be gone.

df_raw["Category"].value_counts(dropna=False)
```

```python
#Handle missing Category

#We have 30 missing categories.

#Since we don't have enough information to determine the actual category, we'll use:Unknown

df_raw["Category"] = df_raw["Category"].fillna("Unknown")
```

```python
df_raw["Category"].isna().sum()
```

```python
df_raw["Category"].value_counts()
```

```python
'''Clean Resolution_Hours and Resolved_Date

First, let's understand the relationship between these columns.

For a ticket:

Resolved → should have Resolution_Hours and Resolved_Date
Open / In Progress / Pending → should normally have these fields blank'''

df_raw.groupby("Status")[[
    "Resolution_Hours",
    "Resolved_Date"
]].count()
```

```python
df_raw.groupby("Status").size()
```

```python
#Check for invalid dates
#We intentionally created some records where:Resolved_Date < Created_Date

invalid_dates = df_raw[
    (df_raw["Resolved_Date"].notna()) &
    (df_raw["Resolved_Date"] < df_raw["Created_Date"])
]

print("Invalid dates:", len(invalid_dates))
```

```python
#"I checked whether the resolved date was earlier than the ticket creation date and identified invalid records."
invalid_dates[
    [
        "Ticket_ID",
        "Created_Date",
        "Resolved_Date",
        "Resolution_Hours",
        "Status"
    ]
].head(10)
```

```python
'''
Fix the invalid dates

We already have Created_Date and Resolution_Hours, so we can correctly reconstruct the Resolved_Date:

Resolved Date = Created Date + Resolution Hours'''

invalid_mask = (
    (df_raw["Status"] == "Resolved") &
    (
        df_raw["Resolved_Date"].isna() |
        (df_raw["Resolved_Date"] < df_raw["Created_Date"])
    )
)

df_raw.loc[invalid_mask, "Resolved_Date"] = (
    df_raw.loc[invalid_mask, "Created_Date"]
    + pd.to_timedelta(
        df_raw.loc[invalid_mask, "Resolution_Hours"],
        unit="hours"
    )
)

print("Invalid/missing resolved dates fixed:", invalid_mask.sum())
```

```python
#Verify the fix

invalid_dates = df_raw[
    (df_raw["Resolved_Date"].notna()) &
    (df_raw["Resolved_Date"] < df_raw["Created_Date"])
]

print("Invalid dates remaining:", len(invalid_dates))
```

```python
missing_resolved_dates = df_raw[
    (df_raw["Status"] == "Resolved") &
    (df_raw["Resolved_Date"].isna())
]

print(
    "Resolved tickets with missing Resolved_Date:",
    len(missing_resolved_dates)
)
```

```python
#Check Resolution Hours by Status
df_raw.groupby("Status")["Resolution_Hours"].count()
```

```python
unresolved_with_hours = df_raw[
    (df_raw["Status"] != "Resolved") &
    (df_raw["Resolution_Hours"].notna())
]

print(
    "Unresolved tickets with Resolution_Hours:",
    len(unresolved_with_hours)
)
```

```python
#Check resolved tickets with missing hours

resolved_missing_hours = df_raw[
    (df_raw["Status"] == "Resolved") &
    (df_raw["Resolution_Hours"].isna())
]

print(
    "Resolved tickets with missing Resolution_Hours:",
    len(resolved_missing_hours)
)
```

```python
#Validate Resolution Hours
#We should also make sure there are no negative resolution times.
negative_hours = df_raw[
    df_raw["Resolution_Hours"] < 0
]

print("Negative Resolution_Hours:", len(negative_hours))
```

```python
#Check the relationship between dates and hours
#For resolved tickets, we can calculate the actual difference between the two dates:

df_raw["Calculated_Resolution_Hours"] = (
    df_raw["Resolved_Date"] - df_raw["Created_Date"]
).dt.total_seconds() / 3600
```

```python
date_difference = (
    df_raw["Calculated_Resolution_Hours"]
    - df_raw["Resolution_Hours"]
).abs()

print(
    "Maximum difference:",
    date_difference.max()
)
```

```python
#Important: Calculated_Resolution_Hours is only a validation column. We don't need it in our final dataset.

df_raw.drop(
    columns=["Calculated_Resolution_Hours"],
    inplace=True
)
```

```python
'''Remove duplicate tickets

Earlier we found 19 exact duplicate rows.

But for a ticket dataset, we should check duplicates using the Ticket_ID, because each ticket should have only one record.'''

#Check duplicate Ticket IDs

duplicate_ticket_ids = df_raw[
    df_raw["Ticket_ID"].duplicated(keep=False)
]

print("Duplicate Ticket IDs:", duplicate_ticket_ids["Ticket_ID"].nunique())
```

```python
duplicate_ticket_ids[
    ["Ticket_ID", "Status", "Category"]
].head(20)
```

```python
#Remove duplicate Ticket IDs
df_raw = df_raw.drop_duplicates(
    subset="Ticket_ID",
    keep="first"
)

print("Shape after removing duplicates:", df_raw.shape)
```

```python
print(
    "Duplicate Ticket IDs remaining:",
    df_raw["Ticket_ID"].duplicated().sum()
)
print("Final rows:", len(df_raw))
```

```python
'''Validate SLA_Status

We need to make sure the SLA status matches the actual resolution time.

Our rule is:

If ticket is not resolved → Not Resolved
If Resolution_Hours <= SLA_Target_Hours → Met
If Resolution_Hours > SLA_Target_Hours → Breached'''
#Recalculate the correct SLA status

calculated_sla_status = np.where(
    df_raw["Resolution_Hours"].isna(),
    "Not Resolved",
    np.where(
        df_raw["Resolution_Hours"] <= df_raw["SLA_Target_Hours"],
        "Met",
        "Breached"
    )
)
```

```python
sla_mismatch = df_raw[
    df_raw["SLA_Status"] != calculated_sla_status
]

print("SLA mismatches:", len(sla_mismatch))
```

```python
#Check SLA by priority
pd.crosstab(
    df_raw["Priority"],
    df_raw["SLA_Status"]
)
```

```python
#Check missing satisfaction values
df_raw["Customer_Satisfaction"].isna().sum()
```

```python
#Check whether any scores are outside 1–5
invalid_satisfaction = df_raw[
    df_raw["Customer_Satisfaction"].notna() &
    (
        (df_raw["Customer_Satisfaction"] < 1) |
        (df_raw["Customer_Satisfaction"] > 5)
    )
]

print("Invalid satisfaction scores:", len(invalid_satisfaction))
```

```python
#Check satisfaction values
df_raw["Customer_Satisfaction"].value_counts().sort_index()
'''Keep missing values as NaN

Do not fill them with 0 or the average.

Missing customer satisfaction means no feedback was provided, not that the customer gave a score of zero.'''
```

```python
#Final data-quality checks
print("Rows:", df_raw.shape[0])
print("Columns:", df_raw.shape[1])

print("\nDuplicate Ticket IDs:")
print(df_raw["Ticket_ID"].duplicated().sum())

print("\nMissing Category:")
print(df_raw["Category"].isna().sum())

print("\nInvalid Dates:")
print(
    (
        df_raw["Resolved_Date"].notna() &
        (df_raw["Resolved_Date"] < df_raw["Created_Date"])
    ).sum()
)

print("\nNegative Resolution Hours:")
print((df_raw["Resolution_Hours"] < 0).sum())
```

```python
#Final check — Status consistency
print(
    "Unresolved tickets with Resolution Hours:",
    (
        (df_raw["Status"] != "Resolved") &
        df_raw["Resolution_Hours"].notna()
    ).sum()
)

print(
    "Resolved tickets without Resolution Hours:",
    (
        (df_raw["Status"] == "Resolved") &
        df_raw["Resolution_Hours"].isna()
    ).sum()
)

print(
    "Resolved tickets without Resolved Date:",
    (
        (df_raw["Status"] == "Resolved") &
        df_raw["Resolved_Date"].isna()
    ).sum()
)
```
